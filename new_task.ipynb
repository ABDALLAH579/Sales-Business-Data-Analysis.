```python
print(df['Category'].value_counts())
print(df['Region'].value_counts())
print(df['Segment'].value_counts())
print(df['Ship Mode'].value_counts())
```

```python
df[['Category', 'Sub-Category', 'Sales']].head()
df[df['Sales'] > 500].head()
df[(df['Sales'] > 500) & (df['Region'] == 'West')].head()
df[(df['Category'] == 'Technology') | (df['Segment'] == 'Corporate')].shape
df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
df.groupby('Region')['Sales'].agg(['sum', 'mean', 'count', 'min', 'max'])
df.groupby(['Category', 'Segment'])['Sales'].sum().unstack()
```

```python
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
df[df['Postal Code'].isnull()][['City', 'State', 'Postal Code']].drop_duplicates()
df.loc[df['City'] == 'Burlington', 'Postal Code'] = 5401
df['Postal Code'] = df['Postal Code'].astype(int)
print(df.isnull().sum().sum())
df = df.drop_duplicates()
print(df.shape)
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d/%m/%Y')
df.dtypes
```

```python
Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df['Sales'] < lower) | (df['Sales'] > upper)]
print(lower, upper)
print(len(outliers), len(outliers) / len(df) * 100)
```

```python
outliers.sort_values('Sales', ascending=False)[['Product Name', 'Category', 'Sales']].head(10)
outliers['Category'].value_counts()
```

```python
sales_by_cat = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
plt.bar(sales_by_cat.index, sales_by_cat.values, color=['#1f77b4', '#ff7f0e', '#2ca02c'], label='Total Sales')
plt.title('Total Sales by Product Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
monthly = df.set_index('Order Date')['Sales'].resample('ME').sum()

plt.figure(figsize=(12, 5))
plt.plot(monthly.index, monthly.values, marker='o', color='teal', label='Monthly Sales')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df[df['Sales'] <= upper]['Sales'], bins=40, color='steelblue', edgecolor='black', label='Sales (without outliers)')
plt.title('Distribution of Sales')
plt.xlabel('Sales Amount')
plt.ylabel('Number of Orders - NOW')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
customers = df.groupby('Customer ID').agg(orders=('Order ID', 'nunique'), total_sales=('Sales', 'sum'))

plt.figure(figsize=(8, 5))
plt.scatter(customers['orders'], customers['total_sales'], alpha=0.5, color='purple', label='Customer')
plt.title('Number of Orders vs Total Sales per Customer')
plt.xlabel('Number of Orders')
plt.ylabel('Total Sales')
plt.legend()
plt.tight_layout()
plt.show()  
```
